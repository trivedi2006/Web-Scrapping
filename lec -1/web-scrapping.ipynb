```python
url1 = "https://subslikescript.com/movies_letter-V"
response = requests.get(url1)
response.status_code
soup1 = BeautifulSoup(response.text,"html.parser")
soup1

```

```python
sc = soup1.find("ul",class_="scripts-list").find_all("a")
sc
links=[]
for i in sc:
    links.append("https://subslikescript.com"+i['href'])
links
```

```python
for  i in links:
    response = requests.get(i)
    response.status_code
    soup = BeautifulSoup(response.text,"html.parser")
    art = soup.find("article",class_="main-article")# ASSIGINING CLASS TO ARTICLE
    script = art.find("div",class_="full-script").text.strip()
    title = art.h1.text
    with open(f"{title}.txt","w") as f :
        f.write(script)
```

```python
for i in range(1,4):
    url = f"https://subslikescript.com/movies_letter-V?page={i}"
    response = requests.get(url)
    response.status_code
    soup = BeautifulSoup(response.text,"html.parser")
    art = soup.find("article",class_="main-article")
    links=[]
    for i in art.find_all("a"):
        links.append("https://subslikescript.com"+i['href'])
    links
    for  i in links:
        try:
            response = requests.get(i)
            response.status_code
            soup = BeautifulSoup(response.text,"html.parser")
            art = soup.find("article",class_="main-article")# ASSIGINING CLASS TO ARTICLE
            script = art.find("div",class_="full-script").text.strip()
            title = art.h1.text
            with open(f"{title}.txt","w") as f :
                f.write(script)  
        except:
            print(i)

        
    
    
    
```

```python
url = "https://webscraper.io/test-sites/e-commerce/allinone/computers/tablets"
res = requests.get(url)
soup=BeautifulSoup(res.text,"html.parser")
box = soup.find_all("div",class_='product-wrapper card-body')
len(box)
```

```python
d = {"name":[],"price":[],"reviews":[]}
for i in box:
    d['name'].append(i.find("a").text.strip())
    d["price"].append(i.find("span").text.strip())
    d["reviews"].append(i.find("p",class_="review-count float-end").text.strip())
df = pd.DataFrame(d)
df
```
