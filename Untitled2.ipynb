```python
import requests
import pandas as pd
from bs4 import BeautifulSoup
import numpy as np
```

```python
with open("Certified used Mercedes-Benz for sale.html","r",encoding="utf-8") as f :
    res = f.read()
soup = BeautifulSoup(res,"html.parser")
```

```python
box = soup.find_all("div",class_="vehicle-details")
print(len(box))
```

```python
d = {'Name':[],'milege':[],'Rating':[],'review':[],'price':[]}
for i in box:
    d['Name'].append(i.find("h2",class_="title").text.split()[1])
    d["milege"].append(i.find("div",class_="mileage").text.split()[0])
    d['price'].append(i.find("div",class_="price-section price-section-vehicle-card").text.split("$")[1].split("\n")[0])
    d['review'].append(i.find("span",class_="sds-rating__link sds-button-link").text.split()[0].split("(")[1])

    try:
            d["Rating"].append(i.find("span",class_="sds-rating__count").text.split()[0])


    except:
        d['Rating'].append(np.nan)
    
df = pd.DataFrame(d)    
df
```
