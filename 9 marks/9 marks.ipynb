```python
import requests
import pandas as pd
from bs4 import BeautifulSoup
import numpy as np
```

```python
with open("demo - Copy (5).html","r",encoding="utf-8") as f :
    res = f.read()
soup = BeautifulSoup(res,"html.parser")
```

```python
box = soup.find_all("div",class_="yKfJKb row")
print(len(box))
```

```python
d = {'Name':[],'ROM':[],'RAM':[],'color':[],'Stars':[],'Reviews':[],'RearCamera':[],'Front Camera':[],'Old_Price':[],'New_Price':[],'discount':[],'rating':[],'display(cm)':[],'display(inch)':[]}
for i in box:
    d['Name'].append(i.find("div",class_="KzDlHZ").text.split("(")[0])
    d["ROM"].append(i.find('ul',class_='G4BRas').find_all('li')[0].text.split("|")[1].split("R")[0].strip())
    d["RAM"].append(i.find('ul',class_='G4BRas').find_all('li')[0].text.split("|")[0])
    d['color'].append(i.find("div",class_="KzDlHZ").text.split("(")[1].split(",")[0])
    d['Stars'].append(float(i.find("div",class_="XQDdHH").text.strip()))
    d['Reviews'].append(int(i.find("span",class_="Wphh3N").text.split('\xa0')[2].split(" ")[0].replace(",","")))
    d['RearCamera'].append(i.find('ul',class_='G4BRas').find_all('li')[2].text.split("|")[0].strip())
    d['Front Camera'].append(i.find('ul',class_='G4BRas').find_all('li')[2].text.split("|")[1].strip().split(" ")[0])
    d['Old_Price'].append(int(i.find('div',class_='Nx9bqj _4b5DiR').text.replace(",","").replace("₹","")))
    try:
        d['New_Price'].append(i.find('div',class_='yRaY8j ZYYwLA').text.replace(",","").replace("₹","")) #Nan
    except:
        d['New_Price'].append(np.nan)
    try:
        d['discount'].append(i.find('div',class_='UkUFwK').text.split(" ")[0]) #Nan
    except:
        d['discount'].append(np.nan)
    d['rating'].append(int(i.find("span",class_="Wphh3N").text.split('\xa0')[0].split(" ")[0].replace(",","")))
    d['display(cm)'].append(float(i.find('ul',class_='G4BRas').find_all('li')[1].text.split(" ")[0]))
    d['display(inch)'].append(float(i.find('ul',class_='G4BRas').find_all('li')[1].text.split(" ")[2].replace("(","")))
df = pd.DataFrame(d)    
df
```
